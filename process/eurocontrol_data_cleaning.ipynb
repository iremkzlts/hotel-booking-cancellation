```python
import pandas as pd #Veri tablolarıyla çalışmamızı sağlayan Pandas kütüphanesini çağırmak.
df_2015 = pd.read_csv('/content/apt_dly_2015.csv') #CSV dosyalarını Pyhton'a okutuyoruz.
df_2016 = pd.read_csv('/content/apt_dly_2016.csv')
df_2017 = pd.read_csv('/content/apt_dly_2017.csv')

print ("2015:", df_2015.shape) # Her dosyanın kaç satır/sütundan oluştuğunu öğreniyoruz.
print ("2016:", df_2016.shape)
print ("2017:", df_2017.shape)
```

```python
#Sütun isimlerini gösterme
print("2015 Sütunları:")
print(df_2015.columns.tolist())
print("\n2016 Sütunları:")
print(df_2016.columns.tolist())
print("\n2017 Sütunları:")
print(df_2017.columns.tolist())


```

```python
# 2016 dosyasındaki sütun isimleri aynı mı kontrol et
print(df_2015.columns.equals(df_2016.columns))

# 2017 dosyasındaki sütun isimleri aynı mı kontrol et
print(df_2015.columns.equals(df_2017.columns))
```

```python
# Üç yıllık veriyi alt alta birleştiriyoruz
df_eurocontrol = pd.concat(
    [df_2015, df_2016, df_2017],
    ignore_index=True
)

# Birleştirilmiş veri setinin boyutunu gösteriyoruz
print("Birleştirilmiş veri boyutu:", df_eurocontrol.shape)

# İlk 5 satırı görüntülüyoruz
df_eurocontrol.head()
```

```python
# Havalimanlarının kodlarına göre kayıt sayılarını gösteriyoruz.
print(df_eurocontrol["APT_ICAO"].value_counts().head(20))
```

```python
# Sadece Faro ve Lisbon havalimanlarının kayıtlarını sayıyoruz
print(
    df_eurocontrol[
        df_eurocontrol["APT_ICAO"].isin(["LPFR", "LPPT"])
    ]["APT_ICAO"].value_counts()
)
```

```python
# Sadece Faro (LPFR) ve Lisbon (LPPT) kayıtlarını alıyoruz
df_eurocontrol_clean = df_eurocontrol[
    df_eurocontrol["APT_ICAO"].isin(["LPFR", "LPPT"])
].copy()
# Filtreleme sonrası veri setinin boyutunu kontrol ediyoruz
print("Filtrelenmiş veri boyutu:", df_eurocontrol_clean.shape)
# Havalimanlarına göre kayıt sayılarını tekrar kontrol ediyoruz
print(df_eurocontrol_clean["APT_ICAO"].value_counts())
```

```python
# FLT_DATE sütununu tarih formatına dönüştürüyoruz
df_eurocontrol_clean["FLT_DATE"] = pd.to_datetime(
    df_eurocontrol_clean["FLT_DATE"]
)

# Tarih sütununun veri tipini kontrol ediyoruz
print(df_eurocontrol_clean["FLT_DATE"].dtype)

# İlk 5 tarihi görüntülüyoruz
print(df_eurocontrol_clean["FLT_DATE"].head())
```

```python
# Tamamen boş olan sütunları buluyoruz
bos_sutunlar = df_eurocontrol_clean.columns[
    df_eurocontrol_clean.isnull().all()
]

# Boş sütunların isimlerini gösteriyoruz
print("Tamamen boş sütunlar:")
print(bos_sutunlar.tolist())

# Kaç tane tamamen boş sütun olduğunu gösteriyoruz
print("Boş sütun sayısı:", len(bos_sutunlar))
```

```python
# Her sütundaki eksik veri sayısını buluyoruz
eksik_veriler = df_eurocontrol_clean.isnull().sum()

# Sadece eksik veri bulunan sütunları gösteriyoruz
print(eksik_veriler[eksik_veriler > 0])

# Her sütundaki eksik veri oranını yüzde olarak hesaplıyoruz
print("\nEksik veri yüzdeleri:")
print((df_eurocontrol_clean.isnull().mean() * 100).round(2))
```

```python
# Gecikme sütunlarındaki eksik verileri inceliyoruz
print("Gecikme bilgisi boş olan kayıtlar:")

print(
    df_eurocontrol_clean[
        df_eurocontrol_clean["DLY_APT_ARR_1"].isnull()
    ][["FLT_DATE", "APT_ICAO", "APT_NAME", "FLT_ARR_1"]].head(10)
)

# Gecikme bilgisi dolu olan kayıtları inceliyoruz
print("\nGecikme bilgisi dolu olan kayıtlar:")

print(
    df_eurocontrol_clean[
        df_eurocontrol_clean["DLY_APT_ARR_1"].notnull()
    ][["FLT_DATE", "APT_ICAO", "APT_NAME", "FLT_ARR_1", "DLY_APT_ARR_1"]].head(10)
)
```

```python
# Gecikme bilgisi içeren sütunları belirliyoruz
gecikme_sutunlari = [
    "DLY_APT_ARR_1",
    "DLY_APT_ARR_A_1",
    "DLY_APT_ARR_C_1",
    "DLY_APT_ARR_D_1",
    "DLY_APT_ARR_E_1",
    "DLY_APT_ARR_G_1",
    "DLY_APT_ARR_I_1",
    "DLY_APT_ARR_M_1",
    "DLY_APT_ARR_N_1",
    "DLY_APT_ARR_O_1",
    "DLY_APT_ARR_P_1",
    "DLY_APT_ARR_R_1",
    "DLY_APT_ARR_S_1",
    "DLY_APT_ARR_T_1",
    "DLY_APT_ARR_V_1",
    "DLY_APT_ARR_W_1",
    "DLY_APT_ARR_NA_1",
    "FLT_ARR_1_DLY",
    "FLT_ARR_1_DLY_15"
]

# Bütün gecikme sütunları boş olan kayıtları sayıyoruz
tamamen_bos_gecikme = df_eurocontrol_clean[
    df_eurocontrol_clean[gecikme_sutunlari].isnull().all(axis=1)
]

print("Bütün gecikme bilgileri boş kayıt sayısı:")
print(len(tamamen_bos_gecikme))

# İlk 10 kaydı görüntülüyoruz
print(tamamen_bos_gecikme[
    ["FLT_DATE", "APT_ICAO", "FLT_ARR_1"]
].head(10))
```

```python
# Gecikme bilgileri tamamen boş olan kayıtları havalimanına göre sayıyoruz
print(tamamen_bos_gecikme["APT_ICAO"].value_counts())

# Eksik gecikme kayıtlarının tarih aralığını buluyoruz
print("\nİlk eksik kayıt tarihi:")
print(tamamen_bos_gecikme["FLT_DATE"].min())

print("\nSon eksik kayıt tarihi:")
print(tamamen_bos_gecikme["FLT_DATE"].max())
```

```python
# Veri setimizin başlangıç ve bitiş tarihlerini buluyoruz
print("İlk tarih:", df_eurocontrol_clean["FLT_DATE"].min())
print("Son tarih:", df_eurocontrol_clean["FLT_DATE"].max())

# Kayıtların yıllara göre dağılımını gösteriyoruz
print("\nYıllara göre kayıt sayısı:")
print(df_eurocontrol_clean["YEAR"].value_counts().sort_index())
```

```python
# Veri setindeki bütün sütunların isimlerini gösteriyoruz
for sutun in df_eurocontrol_clean.columns:
    print(sutun)
```

```python
# Bazı sütunların kaç farklı değer içerdiğini kontrol ediyoruz
kontrol_sutunlari = [
    "APT_ICAO",
    "APT_NAME",
    "STATE_NAME",
    "ATFM_VERSION",
    "YEAR",
    "MONTH_NUM",
    "MONTH_MON"
]

for sutun in kontrol_sutunlari:
    print("\n", sutun)
    print("Farklı değer sayısı:", df_eurocontrol_clean[sutun].nunique())
    print("Değerler:", df_eurocontrol_clean[sutun].unique())
```

```python
# Veri sürümlerinin kaç kayıtta bulunduğunu kontrol ediyoruz
print(df_eurocontrol_clean["ATFM_VERSION"].value_counts(dropna=False))

# Sürümlerin yıllara göre dağılımını inceliyoruz
print("\nYıllara göre veri sürümleri:")
print(
    df_eurocontrol_clean.groupby(
        ["YEAR", "ATFM_VERSION"]
    ).size()
)
```

```python
# Analizimizde kullanmayacağımız veri sürümü sütununu kaldırıyoruz
df_eurocontrol_clean = df_eurocontrol_clean.drop(
    columns=["ATFM_VERSION"]
)

# Güncel veri setimizin boyutunu kontrol ediyoruz
print("Güncel veri boyutu:", df_eurocontrol_clean.shape)
```

```python
# Hava durumu verileriyle ortak tarih aralığını belirliyoruz
baslangic_tarihi = pd.Timestamp("2015-07-01", tz="UTC")
bitis_tarihi = pd.Timestamp("2017-08-31", tz="UTC")

# Belirlediğimiz tarih aralığındaki kayıtları filtreliyoruz
df_eurocontrol_clean = df_eurocontrol_clean[
    (df_eurocontrol_clean["FLT_DATE"] >= baslangic_tarihi) &
    (df_eurocontrol_clean["FLT_DATE"] <= bitis_tarihi)
].copy()

# Filtreleme sonrası veri boyutunu kontrol ediyoruz
print("Yeni veri boyutu:", df_eurocontrol_clean.shape)

# Yeni tarih aralığını kontrol ediyoruz
print("İlk tarih:", df_eurocontrol_clean["FLT_DATE"].min())
print("Son tarih:", df_eurocontrol_clean["FLT_DATE"].max())
```

```python
# Veri setindeki tamamen tekrarlanan kayıtları buluyoruz
tekrarli_kayitlar = df_eurocontrol_clean[
    df_eurocontrol_clean.duplicated()
]

# Tekrarlanan kayıt sayısını gösteriyoruz
print("Tekrarlanan kayıt sayısı:", len(tekrarli_kayitlar))

# Varsa tekrarlanan kayıtları görüntülüyoruz
print(tekrarli_kayitlar.head())
```

```python
# Bütün sütunların veri tiplerini kontrol ediyoruz
print(df_eurocontrol_clean.dtypes)
```

```python
# Uçuş sayısı negatif olan kayıtları kontrol ediyoruz
print("Negatif uçuş sayısı:")
print(df_eurocontrol_clean[df_eurocontrol_clean["FLT_ARR_1"] < 0])

# Gecikme sütunlarındaki negatif değerleri kontrol ediyoruz
gecikme_sutunlari = df_eurocontrol_clean.columns[
    df_eurocontrol_clean.columns.str.startswith("DLY_")
]

print("\nNegatif gecikme değerlerinin sayısı:")
print((df_eurocontrol_clean[gecikme_sutunlari] < 0).sum())
```

```python
# Hiç varış uçuşu olmayan kayıtları buluyoruz
sifir_ucus = df_eurocontrol_clean[
    df_eurocontrol_clean["FLT_ARR_1"] == 0
]

# Sıfır uçuş bulunan kayıt sayısını gösteriyoruz
print("Sıfır uçuş bulunan kayıt sayısı:", len(sifir_ucus))

# Bu kayıtların ilk 10 tanesini görüntülüyoruz
print(
    sifir_ucus[
        ["FLT_DATE", "APT_ICAO", "APT_NAME", "FLT_ARR_1"]
    ].head(10)
)
```

```python
# Gecikme sütunlarını belirliyoruz
gecikme_sutunlari = df_eurocontrol_clean.columns[
    df_eurocontrol_clean.columns.str.startswith("DLY_")
]

# Her gecikme sütunundaki dolu değer sayısını buluyoruz
print("Gecikme sütunlarının dolu değer sayıları:")
print(df_eurocontrol_clean[gecikme_sutunlari].count())

# Her sütundaki farklı değer sayılarını inceliyoruz
print("\nGecikme sütunlarının farklı değer sayıları:")
print(df_eurocontrol_clean[gecikme_sutunlari].nunique())
```

```python
# Tek farklı değere sahip gecikme sütunlarını belirliyoruz
tek_degerli_sutunlar = []

for sutun in gecikme_sutunlari:
    if df_eurocontrol_clean[sutun].nunique() == 1:
        tek_degerli_sutunlar.append(sutun)

# Bu sütunların içerdiği değerleri gösteriyoruz
for sutun in tek_degerli_sutunlar:
    print(sutun, ":", df_eurocontrol_clean[sutun].unique())
```

```python
# Sürekli sıfır değer taşıyan gecikme sütunlarını kaldırıyoruz
df_eurocontrol_clean = df_eurocontrol_clean.drop(
    columns=tek_degerli_sutunlar
)

# Güncel veri boyutunu kontrol ediyoruz
print("Güncel veri boyutu:", df_eurocontrol_clean.shape)
```

```python
# Kalan sütunlardaki boş değer sayılarını hesaplıyoruz
eksik_veriler = df_eurocontrol_clean.isnull().sum()

# Sadece boş değer bulunan sütunları gösteriyoruz
print(eksik_veriler[eksik_veriler > 0])

# Toplam boş hücre sayısını gösteriyoruz
print("\nToplam boş hücre:", df_eurocontrol_clean.isnull().sum().sum())
```

```python
# Gecikme sütunlarının tamamı boş olan kayıtları buluyoruz
tum_gecikmeler_bos = df_eurocontrol_clean[
    df_eurocontrol_clean[gecikme_sutunlari.drop(
        labels=tek_degerli_sutunlar
    )].isnull().all(axis=1)
]

# Bu kayıtların sayısını gösteriyoruz
print("Bütün gecikme bilgileri boş olan kayıt:", len(tum_gecikmeler_bos))

# Örnek kayıtları görüntülüyoruz
print(
    tum_gecikmeler_bos[
        ["FLT_DATE", "APT_ICAO", "APT_NAME", "FLT_ARR_1"]
    ].head()
)

```

```python
# Tarih sütunundan yeniden elde edilebilecek sütunları kaldırıyoruz
df_eurocontrol_clean = df_eurocontrol_clean.drop(
    columns=["YEAR", "MONTH_NUM", "MONTH_MON"]
)

# Sütun adlarını ve veri boyutunu kontrol ediyoruz
print("Son veri boyutu:", df_eurocontrol_clean.shape)
print("\nSon sütunlar:")
print(df_eurocontrol_clean.columns.tolist())

# Son eksik veri kontrolü
print("\nEksik değer sayısı:")
print(df_eurocontrol_clean.isnull().sum())
```

```python
# Temizlenmiş EUROCONTROL verisini CSV dosyası olarak kaydediyoruz
df_eurocontrol_clean.to_csv(
    "/content/eurocontrol_clean.csv",
    index=False
)

# Kaydetme işlemini kontrol ediyoruz
print("EUROCONTROL temiz veri seti kaydedildi!")
print("Satır sayısı:", len(df_eurocontrol_clean))
print("Sütun sayısı:", len(df_eurocontrol_clean.columns))
```

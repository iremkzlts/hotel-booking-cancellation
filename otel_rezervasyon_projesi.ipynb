```python
month_map = {
    "January": 1, "February": 2, "March": 3,
    "April": 4, "May": 5, "June": 6,
    "July": 7, "August": 8, "September": 9,
    "October": 10, "November": 11, "December": 12
}

df_clean["arrival_date"] = pd.to_datetime({
    "year": df_clean["arrival_date_year"],
    "month": df_clean["arrival_date_month"].map(month_map),
    "day": df_clean["arrival_date_day_of_month"]
})

df_clean["reservation_status_date"] = pd.to_datetime(
    df_clean["reservation_status_date"]
)

df_clean["status_arrival_diff"] = (
    df_clean["reservation_status_date"] - df_clean["arrival_date"]
).dt.days

print("REZERVASYON DURUM TARİHİ - VARIŞ TARİHİ KONTROLÜ")

display(
    df_clean.groupby("reservation_status")["status_arrival_diff"]
    .agg(["count", "min", "median", "max"])
)

print("\nDurum tarihinin varış tarihine göre konumu:")

display(
    df_clean.assign(
        date_relation=df_clean["status_arrival_diff"].apply(
            lambda x: "Varıştan önce" if x < 0
            else ("Varış günü" if x == 0 else "Varıştan sonra")
        )
    ).groupby(["reservation_status", "date_relation"])
    .size()
    .unstack(fill_value=0)
)
```
